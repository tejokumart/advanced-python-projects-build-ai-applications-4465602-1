## Create and Train TF-IDF Model

### Subtask:
Initialize a TfidfVectorizer and fit it to the loaded text data. This step builds the vocabulary and learns the term frequencies and inverse document frequencies, which are essential for calculating text similarity.


**Reasoning**:
To initialize the TfidfVectorizer, fit it to the text data, and transform the data into TF-IDF vectors, I need to import the class from scikit-learn and then apply the fit_transform method to the `text_data`.



```python
from sklearn.feature_extraction.text import TfidfVectorizer
import re # Import the regular expression module

def custom_preprocessor(text):
    text = text.lower()
    # Remove 'var_' prefix if present, only if it's at the beginning
    if text.startswith('var_'):
        text = text[4:]

    # Replace all non-alphanumeric characters (except spaces for now) with a single space
    # This handles _, /, ., -, etc.
    text = re.sub(r'[^a-z0-9]+', ' ', text)

    # Remove leading/trailing spaces and multiple spaces
    text = re.sub(r'\s+', ' ', text).strip()

    return text

# Initialize TfidfVectorizer with custom preprocessor
tfidf_vectorizer = TfidfVectorizer(preprocessor=custom_preprocessor)

# Fit and transform the text data
tfidf_matrix = tfidf_vectorizer.fit_transform(text_data)

print("TF-IDF Vectorizer initialized and fitted with custom preprocessor.")
print(f"Shape of TF-IDF matrix: {tfidf_matrix.shape}")
```

## Get User Input and Vectorize

### Subtask:
Prompt the user to enter the input string they want to analyze for similarity. This input string will then be transformed into its TF-IDF vector representation using the previously trained TfidfVectorizer.


**Reasoning**:
To prompt the user for an input string and transform it into a TF-IDF vector, I will use a Python code block. This will involve using the `input()` function to get the string and the previously fitted `tfidf_vectorizer` to convert the string into its TF-IDF representation.



```python
user_input_string = input("Please enter the string you want to analyze for similarity: ")

# Transform the user input string into its TF-IDF vector representation
user_input_tfidf = tfidf_vectorizer.transform([user_input_string])

print(f"User input: '{user_input_string}'")
print(f"Shape of user input TF-IDF vector: {user_input_tfidf.shape}")
```

## Calculate Similarity

### Subtask:
Calculate the cosine similarity between the user's input TF-IDF vector and all the TF-IDF vectors of the `text_data`.

#### Instructions
1. Import the `cosine_similarity` function from `sklearn.metrics.pairwise`.
2. Calculate the cosine similarity between `user_input_tfidf` and `tfidf_matrix`.
3. Store the similarity scores in a variable, for example, `cosine_similarities`.

**Reasoning**:
To calculate the cosine similarity, I need to import the `cosine_similarity` function and then apply it to the `user_input_tfidf` and `tfidf_matrix`.



```python
from sklearn.metrics.pairwise import cosine_similarity

# Calculate cosine similarity between the user input TF-IDF vector and all text data TF-IDF vectors
cosine_similarities = cosine_similarity(user_input_tfidf, tfidf_matrix)

print(f"Shape of cosine similarities: {cosine_similarities.shape}")
```

## Display Most Similar Strings

### Subtask:
Retrieve and display the top N most similar strings from the original `text_data` based on the calculated cosine similarities. The user should be able to specify the number of top similar strings to display.

#### Instructions
1. Prompt the user to enter the number of top similar strings they want to see.
2. Get the indices of the top N most similar strings.
3. Print the top N similar strings along with their similarity scores.

**Reasoning**:
To display the top N most similar strings, I need to prompt the user for the number of results, sort the cosine similarities in descending order, and then retrieve and print the corresponding strings from `text_data` along with their scores.



```python
import numpy as np

try:
    num_similar = int(input("Enter the number of top similar strings to display: "))
except ValueError:
    print("Invalid input. Please enter an integer.")
    num_similar = 5 # Default to 5 if input is invalid

# Flatten the cosine similarities array to easily sort
similarity_scores = cosine_similarities.flatten()

# Get the indices of the top N most similar strings (in descending order)
top_n_indices = similarity_scores.argsort()[-num_similar:][::-1]

print(f"\nTop {num_similar} most similar strings to '{user_input_string}':")
for i, index in enumerate(top_n_indices):
    score = similarity_scores[index]
    original_string = text_data[index]
    print(f"  {i+1}. Score: {score:.4f}, String: '{original_string}'")
```
